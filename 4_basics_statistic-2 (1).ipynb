```python
import numpy as np
import pandas as pd
```

```python
# loading data
patient = pd.read_csv("Patient_Data.csv")
```

```python
# info of dataset
patient.info()
```

```python
#finding nulls
patient.isna().sum()
```

```python
patient.head(10)
```

```python
# statistics
patient.describe()
```

```python
# droping unwanted columns
patient.drop(['ReceptionistID',	'CheckInTime'] , axis=1,inplace=True)
```

```python
# grouping the data by dept
patient.groupby(['Department' , 'BillAmount']).sum()
```

```python
# dropping the duplicates
patient = patient.drop_duplicates(subset=['PatientID'])
patient
```

```python
# filling the null values with mean
patient['BillAmount'] = patient['BillAmount'].fillna(patient['BillAmount'].mean())
```

```python
patient
```

```python
# loding billing dataset
billing = pd.read_csv('Patient_Data.csv')
billing
```

```python
#concatenate
#pd.concat([patient, billing[['InsuranceCovered',	'FinalAmount']]] , axis=1)
```

```python
# merging the two data set on basis of Patient_ID
final_result = pd.merge(patient, billing, on='PatientID')
final_result
```
